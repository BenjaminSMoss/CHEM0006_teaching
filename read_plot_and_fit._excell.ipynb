```python
import pandas as pd # pandas is python's excel like module
import matplotlib.pyplot as plt # matplotlib is python's plotting module
# import X as Y syntax means we can use a shortened name for the module.
# # This is useful to avoid typing out the full name every time.
```

We can now use pandas, the excell module to read out data

```python
data_name='dumm_data.xlsx'# change this line to be 
#the name of your file. Be sure to include '' or "" around the name.
#also be sure to include the file extension (.xlsx, .csv, .txt, etc.).

spreadsheet_data=pd.read_excel(data_name)
# this tells python to read the data from the file into a dataframe.
# the dataframe can be called by its - here we call it spreadsheet_data.
```

Special 'objects' in python come with added functionality that we can access with  "dot notation":

```python
spreadsheet_data.head() # this command prints the first 5 rows of the dataframe.
```

```python
spreadsheet_data.describe() # this command prints a statistical 
```

```python
# summary of the dataframe. Handy right?
spreadsheet_data.plot(x='Mass(ug)',y='Absorbance',kind='scatter')
# the 'kind' argument tells matplotlib to plot a scatter plot.
plt.xlabel('x') #plt is another special object (a pyplot object)
plt.ylabel('y') #  it lets us access and modify the plot 
plt.title('My first plot')
plt.show()
```

```python
from scipy.stats import linregress # 'from X import Y' loads just one function from a module

x = spreadsheet_data['Mass(ug)'] # pick out a column by its header name
y = spreadsheet_data['Absorbance']

fit = linregress(x, y) # this does the linear regression, like Excel's LINEST
# lingress makes a 'fit object' that contains the regression results.
# the attributes of the fit object can be accessed with dot notation:

print(fit.slope)
print(fit.intercept)
print(fit.rvalue**2)
```

```python
spreadsheet_data.plot(x='Mass(ug)',y='Absorbance',kind='scatter')
plt.plot(x, fit.slope * x + fit.intercept, color='red')

# "f string syntax" is used to embed variables in a string. 
title=f'y = {fit.slope:.3f}x + {fit.intercept:.3f}, 'f'R$^2$ = {fit.rvalue**2:.4f}'
plt.xlabel('Mass(ug)') 
plt.ylabel('Absorbance')
plt.title(title)
plt.savefig('dumm_data_plot.png')
```
